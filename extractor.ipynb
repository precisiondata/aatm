```python
results = await extractor([clinical_note])
```

```python
import json

tree = {}
for e in results[0].validated_entities:
    if e.text.casefold() not in tree and e.parents is None:
        tree[e.text.casefold()] = {}

    if e.parents is None:
        tree[e.text.casefold()][e.key] = e.text

    if e.parents:
        for p in e.parents:
            if p.casefold() not in tree:
                tree[p.casefold()] = {}

            tree[p.casefold()][e.key] = e.text

print(json.dumps(tree, indent=2))
```
